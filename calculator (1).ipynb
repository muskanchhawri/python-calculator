```python
history = []


def calculator():
    try:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
    except ValueError:
        print("Please enter numbers only.")
        return

    op = input("Enter operator (+, -, *, /): ")

    if op == "+":
        result = a + b
        print("Result:", result)
        history.append(f"{a} + {b} = {result}")

    elif op == "-":
        result = a - b
        print("Result:", result)
        history.append(f"{a} - {b} = {result}")

    elif op == "*":
        result = a * b
        print("Result:", result)
        history.append(f"{a} * {b} = {result}")

    elif op == "/":
        if b == 0:
            print("Cannot divide by zero.")
        else:
            result = a / b
            print("Result:", result)
            history.append(f"{a} / {b} = {result}")

    else:
        print("Invalid operator.")


while True:
    print("\n====================")
    print("      CALCULATOR")
    print("====================")
    print("1. Calculate")
    print("2. View History")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        calculator()

    elif choice == "2":
        print("\n--- Calculation History ---")

        if len(history) == 0:
            print("No calculations yet.")
        else:
            for item in history:
                print(item)

    elif choice == "3":
        print("Calculator closed. Thank you!")
        break

    else:
        print("Invalid choice. Please select 1, 2, or 3.")
```
